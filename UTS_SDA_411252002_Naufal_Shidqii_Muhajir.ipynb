```python

# SOAL 1: SEARCHING DAN MERGE SORT

# Digit terakhir NIM
d = 2

# Data mahasiswa
data = [
    ['TI001', 'Alya', 78],
    ['TI002', 'Bima', 80 + d],
    ['TI003', 'Citra', 67],
    ['TI004', 'Danu', 92],
    ['TI005', 'Eka', 74],
    ['TI006', 'Fajar', 88],
    ['TI007', 'Gina', 81],
    ['TI008', 'Hadi', 65 + d],
    ['TI009', 'Intan', 91],
    ['TI010', 'Joko', 76],
    ['TI011', 'Karin', 84],
    ['TI012', 'Lukman', 70 + d]
]

# 1. Linear Search
def linear_search(data, target):
    jumlah = 0

    for i in range(len(data)):
        jumlah += 1

        if data[i][0] == target:
            return i, jumlah

    return -1, jumlah


# 2. Menggabungkan dua bagian data
def merge(kiri, kanan):
    hasil = []

    while len(kiri) > 0 and len(kanan) > 0:
        if kiri[0][0] < kanan[0][0]:
            hasil.append(kiri.pop(0))
        else:
            hasil.append(kanan.pop(0))

    hasil += kiri
    hasil += kanan

    return hasil


# 3. Recursive Merge Sort
def merge_sort(data):
    if len(data) <= 1:
        return data

    tengah = len(data) // 2

    kiri = merge_sort(data[:tengah])
    kanan = merge_sort(data[tengah:])

    return merge(kiri, kanan)


# 4. Binary Search
def binary_search(data, target):
    kiri = 0
    kanan = len(data) - 1
    jumlah = 0

    while kiri <= kanan:
        tengah = (kiri + kanan) // 2
        jumlah += 1

        if data[tengah][0] == target:
            return tengah, jumlah

        elif target < data[tengah][0]:
            kanan = tengah - 1

        else:
            kiri = tengah + 1

    return -1, jumlah


# MENJALANKAN PROGRAM

print("DATA AWAL")
for m in data:
    print(m)

# Linear Search
target = "TI006"
hasil, jumlah = linear_search(data, target)

print("\nLINEAR SEARCH")
print("Target:", target)
print("Index:", hasil)
print("Jumlah pemeriksaan:", jumlah)

# Target tidak ditemukan
hasil, jumlah = linear_search(data, "TI999")
print("\nPencarian TI999:", hasil)
print("Jumlah pemeriksaan:", jumlah)

# Merge Sort
data_urut = merge_sort(data.copy())

print("\nDATA SETELAH MERGE SORT")
for m in data_urut:
    print(m)

# Binary Search
hasil, jumlah = binary_search(data_urut, target)

print("\nBINARY SEARCH")
print("Target:", target)
print("Index:", hasil)
print("Jumlah pemeriksaan:", jumlah)

# Target tidak ditemukan
hasil, jumlah = binary_search(data_urut, "TI999")
print("\nPencarian TI999:", hasil)
print("Jumlah pemeriksaan:", jumlah)
```

```python

# SOAL 2: LINKED QUEUE DAN STACK

# 1. Class Node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# 2. Class LinkedQueue
class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def enqueue(self, data):
        baru = Node(data)

        if self.rear is None:
            self.front = baru
        else:
            self.rear.next = baru

        self.rear = baru
        self.count += 1

    def enqueue_front(self, data):
        baru = Node(data)
        baru.next = self.front
        self.front = baru

        if self.rear is None:
            self.rear = baru

        self.count += 1

    def dequeue(self):
        if self.front is None:
            print("Queue kosong!")
            return None

        data = self.front.data
        self.front = self.front.next
        self.count -= 1

        if self.front is None:
            self.rear = None

        return data

    def peek(self):
        if self.front is None:
            return None
        return self.front.data

    def is_empty(self):
        return self.count == 0

    def size(self):
        return self.count

    def display(self):
        sekarang = self.front
        isi = []

        while sekarang:
            isi.append(sekarang.data)
            sekarang = sekarang.next

        print("Queue:", " -> ".join(isi) if isi else "Kosong")


# 3. Class Stack
class Stack:
    def __init__(self):
        self.data = []

    def push(self, data):
        self.data.append(data)

    def pop(self):
        if self.is_empty():
            print("Stack kosong!")
            return None
        return self.data.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.data[-1]

    def is_empty(self):
        return len(self.data) == 0

    def size(self):
        return len(self.data)

    def display(self):
        print("Stack:", " -> ".join(self.data) + " (TOP)"
              if self.data else "Stack: Kosong")


# 4. Membuat Queue dan Stack
queue = LinkedQueue()
stack = Stack()


# 5. Fungsi menampilkan kondisi
def tampilkan():
    queue.display()
    stack.display()
    print("Jumlah Queue:", queue.size())
    print("------------------------")


# 6. Proses DATANG
def datang(nama):
    queue.enqueue(nama)
    print("DATANG:", nama)
    tampilkan()


# 7. Proses LAYANI
def layani():
    nama = queue.dequeue()

    if nama is not None:
        stack.push(nama)
        print("LAYANI:", nama)

    tampilkan()


# 8. Proses UNDO
def undo():
    nama = stack.pop()

    if nama is not None:
        queue.enqueue_front(nama)
        print("UNDO:", nama)

    tampilkan()


# 9. Pengujian dasar Queue
print("PENGUJIAN QUEUE")

uji = LinkedQueue()
uji.enqueue("A")
uji.enqueue("B")
uji.enqueue("C")

uji.display()
print("Front:", uji.peek())
print("Size:", uji.size())
print("Dequeue:", uji.dequeue())

uji.enqueue_front("A")
uji.display()
print("Is Empty:", uji.is_empty())


# 10. Pengujian dasar Stack
print("\nPENGUJIAN STACK")

uji_stack = Stack()
uji_stack.push("A")
uji_stack.push("B")
uji_stack.push("C")

uji_stack.display()
print("Peek:", uji_stack.peek())
print("Pop:", uji_stack.pop())
uji_stack.display()
print("Size:", uji_stack.size())


# 11. Pengujian kondisi kosong
print("\nPENGUJIAN KONDISI KOSONG")

queue_kosong = LinkedQueue()
stack_kosong = Stack()

print("Dequeue Queue kosong:")
queue_kosong.dequeue()

print("Undo Stack kosong:")
stack_kosong.pop()


# 12. SIMULASI 12 OPERASI
print("\nSIMULASI 12 OPERASI")

datang("MHS-501")
datang("MHS-502")
datang("MHS-503")
layani()
datang("MHS-504")
layani()
datang("MHS-505")
undo()
layani()
layani()
undo()
datang("MHS-506")


# 13. HASIL AKHIR
print("\nHASIL AKHIR")

queue.display()
stack.display()

print("Front:", queue.front.data if queue.front else None)
print("Rear:", queue.rear.data if queue.rear else None)
print("Jumlah Queue:", queue.size())
print("Jumlah Stack:", stack.size())
```

2. Fungsi Linear Search

```python
def linear_search(data, target):
    jumlah = 0

    for i in range(len(data)):
        jumlah += 1

        if data[i][0] == target:
            return i, jumlah

    return -1, jumlah
```

Penjelasan:

Linear Search adalah metode untuk mencari data dengan cara memeriksa satu per satu mulai dari data pertama.
- jumlah = 0 digunakan untuk menghitung berapa kali data diperiksa.
- for digunakan untuk memeriksa setiap data.

3. Fungsi Merge

```python
def merge(kiri, kanan):
    hasil = []

    while len(kiri) > 0 and len(kanan) > 0:
        if kiri[0][0] < kanan[0][0]:
            hasil.append(kiri.pop(0))
        else:
            hasil.append(kanan.pop(0))

    hasil += kiri
    hasil += kanan

    return hasil
```

Penjelasan:

Merge digunakan untuk menggabungkan dua bagian data yang sudah diurutkan.
- hasil = [] membuat tempat untuk menyimpan hasil penggabungan.
- while digunakan selama kedua bagian masih memiliki data.
- if membandingkan NIM dari bagian kiri dan kanan.
- Data dengan NIM lebih kecil dimasukkan terlebih dahulu.
- hasil += kiri dan hasil += kanan memasukkan sisa data.
- return hasil mengembalikan data yang sudah digabungkan.

4. Fungsi Merge Sort

```python
def merge_sort(data):
    if len(data) <= 1:
        return data

    tengah = len(data) // 2

    kiri = merge_sort(data[:tengah])
    kanan = merge_sort(data[tengah:])

    return merge(kiri, kanan)
```

Penjelasan:

Merge Sort adalah metode pengurutan data dengan cara membagi data menjadi beberapa bagian kecil, kemudian menggabungkannya kembali dalam urutan yang benar.
- if len(data) <= 1 memeriksa apakah data hanya memiliki satu atau tidak memiliki elemen. Jika iya, data sudah dianggap terurut.
- tengah digunakan untuk mencari posisi tengah data.
- kiri menyimpan bagian pertama.
- kanan menyimpan bagian kedua.
- merge_sort() dipanggil kembali untuk mengurutkan kedua bagian.
- merge() menggabungkan kedua bagian menjadi satu data yang terurut.
Contohnya, data mahasiswa yang awalnya belum terurut akan diurutkan berdasarkan NIM dari TI001 sampai TI012.

5. Fungsi Binary Search

```python
def binary_search(data, target):
    kiri = 0
    kanan = len(data) - 1
    jumlah = 0

    while kiri <= kanan:
        tengah = (kiri + kanan) // 2
        jumlah += 1

        if data[tengah][0] == target:
            return tengah, jumlah

        elif target < data[tengah][0]:
            kanan = tengah - 1

        else:
            kiri = tengah + 1

    return -1, jumlah
```

Penjelasan:

Binary Search adalah metode pencarian data dengan cara membagi area pencarian menjadi dua bagian.
- kiri = 0 menentukan batas awal pencarian.
- kanan menentukan batas akhir pencarian.
- tengah menentukan posisi tengah data.
- Jika NIM ditemukan, program menampilkan posisi dan jumlah pemeriksaan.
- Jika target lebih kecil dari NIM tengah, pencarian berpindah ke bagian kiri.
- Jika target lebih besar, pencarian berpindah ke bagian kanan.
- Jika data tidak ditemukan, hasilnya -1.
Catatan: Binary Search hanya dapat bekerja dengan benar jika data sudah diurutkan.

SOAL 2 — Sistem Antrean dengan Linked Queue dan Stack

A. Penjelasan codingan
1. Membuat Class Node

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

Penjelasan:

Node merupakan bagian kecil yang digunakan untuk menyimpan data dalam Linked List.
- self.data digunakan untuk menyimpan data mahasiswa.
- self.next digunakan untuk menunjuk ke Node berikutnya.
- None berarti belum ada data berikutnya.
Contohnya, Node menyimpan data MHS-501 dan dapat terhubung dengan Node berikutnya.

2. Membuat Class LinkedQueue

```python
class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0
```

Penjelasan:

LinkedQueue digunakan untuk membuat sistem antrean menggunakan Linked List.
- front menunjukkan data paling depan dalam antrean.
- rear menunjukkan data paling belakang.
- count digunakan untuk menghitung jumlah data dalam antrean.

3. Fungsi Enqueue

```python
def enqueue(self, data):
    baru = Node(data)

    if self.rear is None:
        self.front = baru
    else:
        self.rear.next = baru

    self.rear = baru
    self.count += 1
```

Penjelasan:

Enqueue digunakan untuk menambahkan mahasiswa ke bagian belakang antrean.
- Membuat Node baru.
- Jika antrean kosong, Node baru menjadi front.
- Jika antrean tidak kosong, Node baru disambungkan setelah rear.
- rear dipindahkan ke Node baru.
- Jumlah antrean bertambah satu.

4. Fungsi Enqueue Front

```python
def enqueue_front(self, data):
    baru = Node(data)
    baru.next = self.front
    self.front = baru

    if self.rear is None:
        self.rear = baru

    self.count += 1
```

Penjelasan:

Enqueue Front digunakan untuk memasukkan data ke bagian paling depan antrean.

Fungsi ini digunakan ketika mahasiswa melakukan Undo, yaitu mahasiswa yang sebelumnya sudah dilayani dimasukkan kembali ke bagian depan antrean.

5. Fungsi Dequeue

```python
def dequeue(self):
    if self.front is None:
        print("Queue kosong!")
        return None

    data = self.front.data
    self.front = self.front.next
    self.count -= 1

    if self.front is None:
        self.rear = None

    return data
```

Penjelasan:

Dequeue digunakan untuk mengeluarkan mahasiswa dari bagian depan antrean.
- Jika antrean kosong, program menampilkan pesan.
- Jika ada data, program mengambil data paling depan.
- front dipindahkan ke Node berikutnya.
- Jumlah antrean dikurangi satu.
- Jika antrean menjadi kosong, rear juga diatur menjadi None.

6. Fungsi Peek, Is Empty, dan Size

```python
def peek(self):
    if self.front is None:
        return None
    return self.front.data

def is_empty(self):
    return self.count == 0

def size(self):
    return self.count
```

Penjelasan:
- peek() digunakan untuk melihat data paling depan tanpa menghapusnya.
- is_empty() digunakan untuk memeriksa apakah antrean kosong.
- size() digunakan untuk mengetahui jumlah data dalam antrean.

7. Fungsi Display

```python
def display(self):
    sekarang = self.front
    isi = []

    while sekarang:
        isi.append(sekarang.data)
        sekarang = sekarang.next

    print("Queue:", " -> ".join(isi) if isi else "Kosong")
```

Penjelasan:

Display digunakan untuk menampilkan seluruh data mahasiswa yang ada di antrean.
- sekarang dimulai dari bagian depan antrean.
- while digunakan untuk membaca setiap Node.
- isi.append() memasukkan data ke daftar sementara agar mudah ditampilkan.
- sekarang = sekarang.next berpindah ke Node berikutnya.
- Jika tidak ada data, program menampilkan tulisan Kosong.

8. Membuat Class Stack

```python
class Stack:
    def __init__(self):
        self.data = []
```

Penjelasan:

Stack digunakan untuk menyimpan riwayat mahasiswa yang sudah dilayani.

Stack menggunakan prinsip LIFO (Last In, First Out), artinya data yang terakhir masuk akan menjadi data pertama yang keluar.

9. Fungsi Push dan Pop

```python
def push(self, data):
    self.data.append(data)

def pop(self):
    if self.is_empty():
        print("Stack kosong!")
        return None
    return self.data.pop()
```

Penjelasan:
- push() digunakan untuk menambahkan data ke Stack.
- pop() digunakan untuk mengambil dan menghapus data paling atas.
- Jika Stack kosong, program menampilkan pesan dan tidak mengalami error.
Contohnya, jika Stack berisi MHS-501 dan MHS-502, maka pop() akan mengambil MHS-502 terlebih dahulu.

10. Fungsi Peek, Is Empty, Size, dan Display pada Stack

```python
def peek(self):
    if self.is_empty():
        return None
    return self.data[-1]

def is_empty(self):
    return len(self.data) == 0

def size(self):
    return len(self.data)

def display(self):
    print("Stack:", " -> ".join(self.data) + " (TOP)"
          if self.data else "Stack: Kosong")
```

Penjelasan:
- peek() melihat data paling atas tanpa menghapusnya.
- is_empty() memeriksa apakah Stack kosong.
- size() menghitung jumlah data.
- display() menampilkan seluruh isi Stack.

11. Fungsi Datang

```python
def datang(nama):
    queue.enqueue(nama)
    print("DATANG:", nama)
    tampilkan()
```

Penjelasan:

Fungsi datang digunakan ketika mahasiswa baru masuk ke antrean.

Data mahasiswa dimasukkan ke bagian belakang Queue, kemudian program menampilkan kondisi Queue dan Stack.

12. Fungsi Layani

```python
def layani():
    nama = queue.dequeue()

    if nama is not None:
        stack.push(nama)
        print("LAYANI:", nama)

    tampilkan()
```

Penjelasan:

Fungsi layani digunakan ketika mahasiswa mendapatkan pelayanan.
- Mengambil mahasiswa dari bagian depan Queue.
- Mahasiswa yang berhasil dilayani dimasukkan ke Stack sebagai riwayat.
- Program menampilkan kondisi Queue dan Stack.

13. Fungsi Undo

```python
def undo():
    nama = stack.pop()

    if nama is not None:
        queue.enqueue_front(nama)
        print("UNDO:", nama)

    tampilkan()
```

Penjelasan:

Undo digunakan untuk membatalkan pelayanan terakhir.
- Mengambil mahasiswa paling atas dari Stack.
- Mahasiswa tersebut dimasukkan kembali ke bagian depan Queue.
- Program menampilkan kondisi Queue dan Stack.
