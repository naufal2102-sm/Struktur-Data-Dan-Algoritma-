Contoh 2.15 Mencari seluruh kemunculan target

```python
def linear_search_all(data, target):
  posisi = []
  for index, value in enumerate(data):
    if value == target:
      posisi.append(index)
  return posisi

data = [20, 10, 20, 30, 20]
print(linear_search_all(data, 20))
```

2.10.1 Menambahkan Jejak Eksekusi pada Program
Contoh 2.16 Linear Search dengan jejak eksekusi

```python
def linear_search_trace(data, target):
  for index, value in enumerate(data):
    print('Periksa indeks', index, 'nilai', value)
    if value == target:
      print('Target ditemukan')
      return index
  print('Target tidak ditemukan')
  return -1

linear_search_trace([7, 11, 15, 18, 25, 30], 25)
```

2.12. Implementasi Binary Search dengan Python
Contoh 2.17 Fungsi Binary Search iteratif

```python
def binary_search(data, target):
  left = 0
  right = len(data) - 1

  while left <= right:
    mid = (left + right) // 2

    if data[mid] == target:
      return mid
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1

return -1
```

2.12.2 Menjalankan Fungsi
Contoh 2.18 Pengujian Binary Search

```python
data = [12, 18, 25, 31, 44, 57, 63, 70, 82]

print(binary_search(data, 57))
print(binary_search(data, 13))
```

2.13.1 Versi dengan Jejak Eksekusi
Contoh 2.19 Binary Search dengan trace

```python
def binary_search_trace(data, target):
  left = 0
  right = len(data) - 1

  while left <= right:
    mid = (left + right) // 2
    print('left=', left, 'right=', right, 'mid=', mid, 'nilai=', data[mid])

    if data[mid] == target:
      return mid
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1

  return -1

binary_search_trace([12,18,25,31,44,57,63,70,82], 57)
```

Contoh 2.20 Eksperimen sederhana menggunakan timeit

```python
import timeit

data = list(range(100_000))
target = -1

waktu_linear = timeit.timeit(
lambda: linear_search(data, target),
number=100
)

waktu_binary = timeit.timeit(
lambda: binary_search(data, target),
number=100
)

print('Linear Search:', waktu_linear)
print('Binary Search:', waktu_binary)
```

Contoh 2.21 Mengamati pengaruh ukuran input

```python
for n in [1_000, 10_000, 100_000]:
  data = list(range(n))
  target = -1

  t_linear = timeit.timeit(
    lambda: linear_search(data, target),
    number=100
)
  t_binary = timeit.timeit(
    lambda: binary_search(data, target),
    number=100
)

print('n =', n)
print(' linear:', t_linear)
print(' binary:', t_binary)
```
