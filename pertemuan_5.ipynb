5.13. Queue Menggunakan Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def is_empty(self):
        return self.front is None

    def enqueue(self, data):
        new_node = Node(data)

        if self.is_empty():
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

        self.count += 1

    def dequeue(self):
        if self.is_empty():
            return None

        data = self.front.data
        self.front = self.front.next
        self.count -= 1

        if self.front is None:
            self.rear = None

        return data

    def peek(self):
        return None if self.is_empty() else self.front.data

    def size(self):
        return self.count
```

5.16. Implementasi Circular Queue Berkapasitas Tetap

```python
class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.items = [None] * capacity
        self.front = 0
        self.rear = 0
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def is_full(self):
        return self.count == self.capacity

    def enqueue(self, data):
        if self.is_full():
            return False

        self.items[self.rear] = data
        self.rear = (self.rear + 1) % self.capacity
        self.count += 1
        return True

    def dequeue(self):
        if self.is_empty():
            return None

        data = self.items[self.front]
        self.items[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        return data

    def peek(self):
        if self.is_empty():
            return None
        return self.items[self.front]

    def size(self):
        return self.count
```

5.20. Studi Kecil Deque: Pemeriksaan Palindrome

```python
from collections import deque

def is_palindrome(text):
    chars = deque(c.lower() for c in text if c.isalnum())

    while len(chars) > 1:
        if chars.popleft() != chars.pop():
            return False

    return True


print(is_palindrome('Katak'))
print(is_palindrome('Struktur Data'))
```

5.22. Mini Studi Kasus — Sistem Antrean Layanan

```python
from collections import deque

class ServiceQueue:
    def __init__(self):
        self.queue = deque()
        self.next_number = 1

    def arrive(self, name):
        ticket = f'A{self.next_number:03d}'
        self.next_number += 1
        self.queue.append((ticket, name))
        return ticket

    def serve(self):
        if not self.queue:
            return None
        return self.queue.popleft()

    def waiting_count(self):
        return len(self.queue)

    def next_customer(self):
        return None if not self.queue else self.queue[0]


service = ServiceQueue()

print(service.arrive('Ani'))
print(service.arrive('Budi'))
print(service.arrive('Citra'))

print('Berikutnya:', service.next_customer())
print('Dilayani :', service.serve())
print('Menunggu :', service.waiting_count())
```

5.25. Praktik Mandiri 5.1 — Queue Dasar
1. Membuat class Queue

```python
from collections import deque

class Queue:
    def __init__(self):
        self.items = deque()

    def enqueue(self, data):
        self.items.append(data)

    def dequeue(self):
        if len(self.items) == 0:
            return None
        return self.items.popleft()

    def peek(self):
        if len(self.items) == 0:
            return None
        return self.items[0]

    def size(self):
        return len(self.items)

    def is_empty(self):
        return len(self.items) == 0


# Pengujian Queue
q = Queue()

# Memasukkan data 10, 20, 30, 40, 50
for data in [10, 20, 30, 40, 50]:
    q.enqueue(data)

print("Isi Queue:", list(q.items))
print("Front:", q.peek())
print("Size:", q.size())

# Dequeue dua kali
print("Dequeue pertama:", q.dequeue())
print("Dequeue kedua:", q.dequeue())
print("Isi Queue:", list(q.items))

# Menambahkan 60 dan 70
q.enqueue(60)
q.enqueue(70)
print("Setelah enqueue 60 dan 70:", list(q.items))

# Mengeluarkan seluruh elemen sampai Queue kosong
print("\nMengeluarkan seluruh elemen:")
while not q.is_empty():
    print(q.dequeue())

# Menguji dequeue dan peek ketika Queue kosong
print("\nDequeue saat kosong:", q.dequeue())
print("Peek saat kosong:", q.peek())
print("Apakah Queue kosong?", q.is_empty())
```

5.26. Praktik Mandiri 5.2 — Circular Queue
1. Membuat CircularQueue dengan kapasitas 4

```python
class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.items = [None] * capacity
        self.front = 0
        self.rear = 0
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def is_full(self):
        return self.count == self.capacity

    def enqueue(self, data):
        if self.is_full():
            print("Queue penuh! Data", data, "tidak dapat ditambahkan.")
            return False

        self.items[self.rear] = data
        self.rear = (self.rear + 1) % self.capacity
        self.count += 1
        return True

    def dequeue(self):
        if self.is_empty():
            print("Queue kosong!")
            return None

        data = self.items[self.front]
        self.items[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        return data

    def display(self):
        print("Items :", self.items)
        print("Front :", self.front)
        print("Rear  :", self.rear)
        print("Count :", self.count)
        print("Penuh :", self.is_full())
        print("Kosong:", self.is_empty())
        print("-" * 35)


# Pengujian CircularQueue
cq = CircularQueue(4)

print("Enqueue A")
cq.enqueue("A")
cq.display()

print("Enqueue B")
cq.enqueue("B")
cq.display()

print("Enqueue C")
cq.enqueue("C")
cq.display()

print("Enqueue D")
cq.enqueue("D")
cq.display()

print("Mencoba enqueue E saat penuh")
cq.enqueue("E")
cq.display()

print("Dequeue pertama:", cq.dequeue())
cq.display()

print("Dequeue kedua:", cq.dequeue())
cq.display()

print("Enqueue E")
cq.enqueue("E")
cq.display()

print("Enqueue F")
cq.enqueue("F")
cq.display()
```

5.27. Praktik Mandiri 5.3 — Deque dan Palindrome
1. Membuat fungsi is_palindrome()

```python
from collections import deque

def is_palindrome(text, tampilkan_pasangan=False):
    # Normalisasi: huruf kecil dan hanya huruf/angka
    chars = deque(
        c.lower() for c in text if c.isalnum()
    )

    while len(chars) > 1:
        kiri = chars.popleft()
        kanan = chars.pop()

        if tampilkan_pasangan:
            print(f"Kiri: {kiri} | Kanan: {kanan}")

        if kiri != kanan:
            if tampilkan_pasangan:
                print("Hasil: Bukan palindrome")
            return False

    if tampilkan_pasangan:
        print("Hasil: Palindrome")

    return True


# Pengujian kata
kata = ["level", "katak", "radar", "python", "algoritma"]

for teks in kata:
    print(teks, "=", is_palindrome(teks))

# Pengujian kalimat yang mengandung spasi dan tanda baca
kalimat = "A man, a plan, a canal: Panama!"
print(kalimat, "=", is_palindrome(kalimat))
```
